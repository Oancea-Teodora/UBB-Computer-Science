```python
import torch
import torch.nn as nn
import torch.optim as optim
from collections import OrderedDict
```

```python
# your code here
#model1 = nn.Sequential(OrderedDict([
#    ('hidden', nn.
#]))

# Model1: 
model1 = nn.Sequential(OrderedDict([
    ('hidden', nn.Linear(2, 2)),
    ('hidden_activation', nn.Sigmoid()),
    ('output', nn.Linear(2, 2))
]))

# Model2: 
model2 = nn.Sequential(OrderedDict([
    ('hidden', nn.Linear(2, 5)),
    ('hidden_activation', nn.ReLU()),
    ('output', nn.Linear(5, 2))
]))

# Model3: 
model3 = nn.Sequential(OrderedDict([
    ('hidden1', nn.Linear(2, 4)),
    ('relu1', nn.ReLU()),
    ('hidden2', nn.Linear(4, 3)),
    ('relu2', nn.ReLU()),
    ('output', nn.Linear(3, 2))
]))

```

```python
print(model1)
print(model2)
print(model3)
```

```python
# your code here
#data_in = torch.tensor( ...
data_in = torch.tensor([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=torch.float32)
print(data_in)
```

```python
# your code here
# data_target = torch.tensor( ...
data_target = torch.tensor([
    [0, 0],
    [0, 1],
    [0, 1],
    [1, 0]
], dtype=torch.float32)
print(data_target)
```

```python
# your code here
# criterion = 
# optimizer = 
criterion = nn.BCEWithLogitsLoss()

optimizer1 = optim.SGD(model1.parameters(), lr=0.1)
optimizer2 = optim.SGD(model2.parameters(), lr=0.1)
optimizer3 = optim.SGD(model3.parameters(), lr=0.1)
```

```python
# your code here
# Train the model
num_epochs = 1000

print("1st model:")
for epoch in range(num_epochs):
    outputs_pred = model1(data_in)
    loss = criterion(outputs_pred, data_target)

    optimizer1.zero_grad()
    loss.backward()
    optimizer1.step()

    if (epoch+1) % 100 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')

print("\n2nd model:")
for epoch in range(num_epochs):
    outputs_pred = model2(data_in)
    loss = criterion(outputs_pred, data_target)
    
    optimizer2.zero_grad()
    loss.backward()
    optimizer2.step()
    
    if (epoch+1) % 100 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')

print("\n3rd model:")
for epoch in range(num_epochs):
    outputs_pred = model3(data_in)
    loss = criterion(outputs_pred, data_target)
    
    optimizer3.zero_grad()
    loss.backward()
    optimizer3.step()
    
    if (epoch+1) % 100 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')
```

```python
# your code here
# visualize the resuts
with torch.no_grad():
    outputs = model1(data_in)
    predicted = (outputs >= 0.5).float()
    accuracy = (predicted == data_target).float().mean()
    print(f'\n1st model training accuracy: {accuracy.item():.4f}')
    
with torch.no_grad():
    outputs = model2(data_in)
    predicted = (outputs >= 0.5).float()
    accuracy = (predicted == data_target).float().mean()
    print(f'2nd model training accuracy: {accuracy.item():.4f}')
    
with torch.no_grad():
    outputs = model3(data_in)
    predicted = (outputs >= 0.5).float()
    accuracy = (predicted == data_target).float().mean()
    print(f'3rd model training accuracy: {accuracy.item():.4f}')
```

```python
# your code here
# print model wights
print("\n1st model:")
for name, param in model1.named_parameters():
    if param.requires_grad:
        print(name, param.data)
        
print("\n2nd model:")
for name, param in model2.named_parameters():
    if param.requires_grad:
        print(name, param.data)
        
print("\n3rd model:")
for name, param in model3.named_parameters():
    if param.requires_grad:
        print(name, param.data)
```
